```python
customers   = pd.read_csv('/content/olist_customers_dataset.csv')
geolocation = pd.read_csv('/content/olist_geolocation_dataset.csv')
items       = pd.read_csv('/content/olist_order_items_dataset.csv')
payments    = pd.read_csv('/content/olist_order_payments_dataset.csv')
reviews     = pd.read_csv('/content/olist_order_reviews_dataset.csv')
orders      = pd.read_csv('/content/olist_orders_dataset.csv')
products    = pd.read_csv('/content/olist_products_dataset.csv')
sellers     = pd.read_csv('/content/olist_sellers_dataset.csv')
translation = pd.read_csv('/content/product_category_name_translation.csv')

# aage ke cells isi dictionary ko use karte hain
dfs = {'customers': customers, 'geolocation': geolocation, 'items': items,
       'payments': payments, 'reviews': reviews, 'orders': orders,
       'products': products, 'sellers': sellers, 'translation': translation}

for name, df in dfs.items():
    print(f'{name:12s} -> {df.shape[0]:>9,} rows | {df.shape[1]:>2} cols')
```

```python
# ------------------------------------------------------------
# Step 4.1: Data Cleaning - convert date columns to datetime
# ------------------------------------------------------------
# Dates are stored as plain text in the CSV files. We convert them
# to datetime so we can calculate delivery time, delays, and monthly trends.

# Orders table: all timestamp columns
order_date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for col in order_date_cols:
    orders[col] = pd.to_datetime(orders[col])   # missing values become NaT (not a time)

# Items table: shipping deadline for the seller
items['shipping_limit_date'] = pd.to_datetime(items['shipping_limit_date'])

# Reviews table: when the review was created and answered
reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'])
reviews['review_answer_timestamp'] = pd.to_datetime(reviews['review_answer_timestamp'])

# Verify that the conversion worked
print(orders.dtypes)
```

```python
# ------------------------------------------------------------
# Step 4.2: Data Cleaning - product category translation
# ------------------------------------------------------------
# Category names in the products table are in Portuguese.
# We join the translation table to get English names.

products = products.merge(translation, on='product_category_name', how='left')

# Build one clean 'category' column:
# 1) use the English name when available
# 2) otherwise fall back to the original Portuguese name
#    (a few categories are missing from the translation table)
# 3) if the category itself is missing, label it 'unknown'
products['category'] = (
    products['product_category_name_english']
    .fillna(products['product_category_name'])
    .fillna('unknown')
)

# Make names readable: 'health_beauty' -> 'Health Beauty'
products['category'] = products['category'].str.replace('_', ' ').str.title()

print('Unique categories:', products['category'].nunique())
products[['product_id', 'category']].head()
```

```python
# ------------------------------------------------------------
# Step 4.3: Data Cleaning - geolocation table
# ------------------------------------------------------------
# The geolocation table has many rows for the same zip code
# (about 1 million rows in total). To join it with customers or
# sellers later, we need exactly ONE row per zip code.
# We take the average latitude and longitude of each zip code.

geo = (
    geolocation
    .groupby('geolocation_zip_code_prefix')
    .agg(lat=('geolocation_lat', 'mean'),
         lng=('geolocation_lng', 'mean'))
    .reset_index()
)

print('Geolocation rows before:', f'{len(geolocation):,}')
print('Geolocation rows after :', f'{len(geo):,}')
geo.head()
```

```python
# ------------------------------------------------------------
# Step 4.4: Data Cleaning - order status and date range
# ------------------------------------------------------------
# Check what statuses exist and which time period the data covers.
# This helps us decide which orders to include in the analysis.

print(orders['order_status'].value_counts())

print('\nDate range:',
      orders['order_purchase_timestamp'].min(), 'to',
      orders['order_purchase_timestamp'].max())
```

```python
# ------------------------------------------------------------
# Step 5.1: Build the order-level table (1 row = 1 order)
# ------------------------------------------------------------
# One order can have many items, many payments and sometimes many reviews.
# To avoid duplicated rows after merging, we first summarize each table
# so that it has exactly ONE row per order_id.

# Items: number of items, total product price, total freight (shipping cost)
order_items = (
    items.groupby('order_id')
    .agg(items_count=('order_item_id', 'count'),
         price=('price', 'sum'),
         freight=('freight_value', 'sum'))
    .reset_index()
)

# Payments: total paid, max installments, most used payment type
order_pay = (
    payments.groupby('order_id')
    .agg(payment_value=('payment_value', 'sum'),
         installments=('payment_installments', 'max'),
         payment_type=('payment_type', lambda x: x.mode().iloc[0]))
    .reset_index()
)

# Reviews: average score per order (a few orders have more than one review)
order_rev = (
    reviews.groupby('order_id')
    .agg(review_score=('review_score', 'mean'))
    .reset_index()
)

# Merge everything into one table on order_id / customer_id
orders_df = (
    orders
    .merge(customers, on='customer_id', how='left')
    .merge(order_items, on='order_id', how='left')
    .merge(order_pay, on='order_id', how='left')
    .merge(order_rev, on='order_id', how='left')
)

print('orders_df shape:', orders_df.shape)
print('Duplicate order_id:', orders_df['order_id'].duplicated().sum())
orders_df.head(3)
```

```python
# ------------------------------------------------------------
# Step 5.2: Feature engineering - create new useful columns
# ------------------------------------------------------------

# --- Time features (for trend and pattern analysis) ---
orders_df['purchase_month'] = orders_df['order_purchase_timestamp'].dt.to_period('M').dt.to_timestamp()
orders_df['purchase_year']  = orders_df['order_purchase_timestamp'].dt.year
orders_df['purchase_dow']   = orders_df['order_purchase_timestamp'].dt.day_name()   # Monday, Tuesday, ...
orders_df['purchase_hour']  = orders_df['order_purchase_timestamp'].dt.hour

# --- Delivery features (for delivery performance analysis) ---
# Actual delivery time: purchase -> delivered to customer
orders_df['delivery_days'] = (
    orders_df['order_delivered_customer_date'] - orders_df['order_purchase_timestamp']
).dt.days

# Promised delivery time: purchase -> estimated delivery date
orders_df['estimated_days'] = (
    orders_df['order_estimated_delivery_date'] - orders_df['order_purchase_timestamp']
).dt.days

# Delay: positive = delivered AFTER the estimated date, negative = delivered early
orders_df['delay_days'] = (
    orders_df['order_delivered_customer_date'] - orders_df['order_estimated_delivery_date']
).dt.days

# Flag late deliveries (True if delivered after the estimated date)
orders_df['is_late'] = orders_df['delay_days'] > 0

# Rounded review score (1 to 5) for grouping in charts
orders_df['review_int'] = orders_df['review_score'].round()

orders_df[['order_id', 'purchase_month', 'delivery_days',
           'estimated_days', 'delay_days', 'is_late']].head()
```

```python
# ------------------------------------------------------------
# Step 5.3: Build the item-level table (1 row = 1 item sold)
# ------------------------------------------------------------
# This table is used for category, seller and product-level analysis.
# Base table = items (112,650 rows). We attach order, product and
# seller information to every item.

items_df = (
    items
    # Order info: customer, status, dates, review and delivery details
    .merge(orders_df[['order_id', 'customer_unique_id', 'customer_state', 'customer_city',
                      'order_status', 'order_purchase_timestamp', 'purchase_month',
                      'review_score', 'delivery_days', 'is_late']],
           on='order_id', how='left')
    # Product info: English category name, weight, number of photos
    .merge(products[['product_id', 'category', 'product_weight_g', 'product_photos_qty']],
           on='product_id', how='left')
    # Seller info: seller city and state
    .merge(sellers, on='seller_id', how='left')
)

print('items_df shape:', items_df.shape)
items_df.head(3)
```

```python
# ------------------------------------------------------------
# Step 5.4: Create analysis subsets
# ------------------------------------------------------------
# Not every order should be used for every analysis, so we create
# three focused subsets.

# 1) Valid orders: exclude 'canceled', 'unavailable' and 'created'
#    because no real sale happened (or it is not confirmed yet).
#    Used for revenue, orders and customer analysis.
valid_status = ['delivered', 'shipped', 'approved', 'invoiced', 'processing']
orders_valid = orders_df[orders_df['order_status'].isin(valid_status)].copy()
items_valid  = items_df[items_df['order_status'].isin(valid_status)].copy()

# 2) Delivered orders: only orders that reached the customer and have
#    a delivery date. Used for delivery time, delay and review analysis.
delivered = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['delivery_days'].notna())
].copy()

# 3) Trend mask: keep only complete months (Jan 2017 to Aug 2018).
#    2016 has very few orders and Sep/Oct 2018 data is incomplete,
#    so including them would make the trend look wrong.
trend_mask = (
    (orders_valid['purchase_month'] >= '2017-01-01') &
    (orders_valid['purchase_month'] <= '2018-08-01')
)

print(f'Valid orders: {len(orders_valid):,} | Delivered: {len(delivered):,}')
```

```python
# ------------------------------------------------------------
# Step 6.3: EDA - monthly orders and revenue trend
# ------------------------------------------------------------
# Only complete months are used (Jan 2017 to Aug 2018) via trend_mask.

# Orders per month
monthly = (
    orders_valid[trend_mask]
    .groupby('purchase_month')
    .agg(orders=('order_id', 'nunique'))
    # Revenue per month (from the item-level table, same date range)
    .join(
        items_valid[items_valid['purchase_month'].between('2017-01-01', '2018-08-01')]
        .groupby('purchase_month')['price'].sum()
        .rename('revenue')
    )
)

# Two charts side by side: orders (left) and revenue (right)
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

ax[0].plot(monthly.index, monthly['orders'], marker='o', color='#440154')
ax[0].set_title('Monthly Orders')
ax[0].set_ylabel('Orders')

ax[1].plot(monthly.index, monthly['revenue'], marker='o', color='#21918c')
ax[1].set_title('Monthly Revenue (BRL)')
ax[1].set_ylabel('Revenue')

for a in ax:
    a.tick_params(axis='x', rotation=45)   # rotate month labels so they don't overlap
save('02_monthly_trend')

# Month-over-month (MoM) revenue growth in %
monthly['revenue_mom_%'] = monthly['revenue'].pct_change() * 100

print('Peak month:', monthly['orders'].idxmax().strftime('%b %Y'),
      '-', monthly['orders'].max(), 'orders')
monthly.tail(8)
```

```python
# ============================================================
# REMAINING ANALYSIS - run this cell once, after Cell 26
# ============================================================

# ------------------------------------------------------------
# 6.9 Customer reviews and the impact of delivery
# ------------------------------------------------------------
fig, ax = plt.subplots(1, 3, figsize=(18, 5))

# Review score distribution
sns.countplot(x=delivered['review_int'].dropna().astype(int), ax=ax[0], palette='viridis')
ax[0].set_title('Review Score Distribution')
ax[0].set_xlabel('Score')

# Average review: on-time vs late orders
sns.barplot(x=delivered['is_late'].map({True: 'Late', False: 'On time'}),
            y=delivered['review_score'], ax=ax[1], palette=['#21918c', '#d62728'])
ax[1].set_title('Avg Review: On-time vs Late')
ax[1].set_xlabel('')
ax[1].set_ylabel('Avg score')

# Delivery days for each review score
sns.boxplot(x=delivered['review_int'].dropna().astype(int),
            y=delivered['delivery_days'].clip(upper=60), ax=ax[2], palette='viridis')
ax[2].set_title('Delivery Days by Review Score')
ax[2].set_xlabel('Score')
ax[2].set_ylabel('Days')
save('12_reviews_vs_delivery')

print(delivered.groupby('is_late')['review_score'].mean()
      .rename({True: 'Late', False: 'On time'}).round(2))

# ------------------------------------------------------------
# 6.9b Review score by how many days late the order was
# ------------------------------------------------------------
bins = [-100, 0, 3, 7, 14, 1000]
labels = ['On time/Early', '1-3 days late', '4-7 days late', '8-14 days late', '15+ days late']
delivered['delay_bucket'] = pd.cut(delivered['delay_days'], bins=bins, labels=labels)

d = delivered.groupby('delay_bucket')['review_score'].mean()
plt.figure(figsize=(10, 4.5))
sns.barplot(x=d.index, y=d.values, palette='RdYlGn_r')
plt.title('Review Score drops as Delay increases')
plt.ylabel('Avg Review Score')
plt.xlabel('')
plt.xticks(rotation=15)
save('13_delay_vs_review')

# ------------------------------------------------------------
# 6.10 Freight (shipping cost) as % of product price, by state
# ------------------------------------------------------------
fr = items_valid.groupby('customer_state').agg(price=('price', 'sum'),
                                               freight=('freight_value', 'sum'))
fr['freight_pct'] = fr['freight'] / fr['price'] * 100
fr = fr.sort_values('freight_pct', ascending=False)

plt.figure(figsize=(12, 5))
sns.barplot(x=fr.index, y=fr['freight_pct'], palette='magma')
plt.title('Freight Cost as % of Product Price, by State')
plt.ylabel('Freight / Price (%)')
save('14_freight_by_state')

# ------------------------------------------------------------
# 6.11 Seller analysis (Pareto: how concentrated is revenue?)
# ------------------------------------------------------------
seller = (
    items_valid.groupby('seller_id')
    .agg(revenue=('price', 'sum'),
         orders=('order_id', 'nunique'),
         avg_review=('review_score', 'mean'))
    .sort_values('revenue', ascending=False)
)
seller['cum_share'] = seller['revenue'].cumsum() / seller['revenue'].sum() * 100
seller['seller_rank_pct'] = np.arange(1, len(seller) + 1) / len(seller) * 100

plt.figure(figsize=(9, 5))
plt.plot(seller['seller_rank_pct'], seller['cum_share'], color='#440154', lw=2)
plt.axvline(20, color='red', ls='--', alpha=0.6)   # top 20% of sellers
plt.axhline(80, color='red', ls='--', alpha=0.6)   # 80% of revenue (Pareto reference)
plt.title('Seller Revenue Concentration (Pareto)')
plt.xlabel('% of Sellers (ranked by revenue)')
plt.ylabel('Cumulative % of Revenue')
save('15_seller_pareto')

top20 = seller[seller['seller_rank_pct'] <= 20]['revenue'].sum() / seller['revenue'].sum() * 100
print(f'Top 20% sellers generate {top20:.1f}% of revenue | Total sellers: {len(seller):,}')

# Top 10 sellers by revenue
t = seller.head(10).reset_index()
t['seller_id'] = t['seller_id'].str[:8]            # shorten the long IDs for the chart
plt.figure(figsize=(11, 5))
sns.barplot(x='revenue', y='seller_id', data=t, orient='h')
plt.title('Top 10 Sellers by Revenue')
plt.xlabel('Revenue (BRL)')
plt.ylabel('Seller ID (short)')
save('16_top_sellers')

# ------------------------------------------------------------
# 6.12 Product photos vs sales and price
# ------------------------------------------------------------
ph = (
    items_valid.dropna(subset=['product_photos_qty'])
    .assign(photos=lambda d: d['product_photos_qty'].clip(upper=8))   # 8 = 8 or more photos
    .groupby('photos')
    .agg(items_sold=('order_id', 'count'), avg_price=('price', 'mean'))
)

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
sns.barplot(x=ph.index.astype(int), y=ph['items_sold'], ax=ax[0], color='#21918c')
ax[0].set_title('Items Sold by Number of Product Photos')
ax[0].set_xlabel('Photos (8 = 8+)')
sns.barplot(x=ph.index.astype(int), y=ph['avg_price'], ax=ax[1], color='#440154')
ax[1].set_title('Avg Price by Number of Product Photos')
ax[1].set_xlabel('Photos (8 = 8+)')
save('17_photos_analysis')

# ------------------------------------------------------------
# 7.1 New vs repeat customers
# ------------------------------------------------------------
cust_orders = orders_valid.groupby('customer_unique_id')['order_id'].nunique()
repeat_rate = (cust_orders > 1).mean() * 100

freq = cust_orders.clip(upper=4).value_counts().sort_index()
freq.index = ['1 order', '2 orders', '3 orders', '4+ orders'][:len(freq)]

plt.figure(figsize=(8, 4.5))
ax = sns.barplot(x=freq.index, y=freq.values, palette='viridis')
for i, v in enumerate(freq.values):
    ax.text(i, v, f'{v:,}', ha='center', va='bottom')
plt.title('Customers by Number of Orders')
plt.ylabel('Customers')
save('18_repeat_customers')
print(f'Repeat customer rate: {repeat_rate:.2f}%')

# ------------------------------------------------------------
# 7.2 RFM segmentation
# Recency = days since last purchase | Frequency = number of orders
# Monetary = total amount spent
# ------------------------------------------------------------
rfm_base = orders_valid[orders_valid['order_status'] == 'delivered'].dropna(subset=['payment_value'])
snapshot = rfm_base['order_purchase_timestamp'].max() + pd.Timedelta(days=1)   # "today" for the analysis

rfm = (
    rfm_base.groupby('customer_unique_id')
    .agg(recency=('order_purchase_timestamp', lambda x: (snapshot - x.max()).days),
         frequency=('order_id', 'nunique'),
         monetary=('payment_value', 'sum'))
    .reset_index()
)

# Scores: R and M use 5 equal groups (rank first to avoid duplicate bin edges).
# F uses manual bins because most customers have only 1 order.
rfm['R'] = pd.qcut(rfm['recency'].rank(method='first'), 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm['M'] = pd.qcut(rfm['monetary'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm['F'] = np.select([rfm['frequency'] == 1, rfm['frequency'] == 2], [1, 2], default=3)

# Segment rules (the first matching condition wins)
conditions = [
    (rfm['F'] >= 2),                                          # bought more than once
    (rfm['F'] == 1) & (rfm['R'] >= 4) & (rfm['M'] >= 4),      # recent and big spender
    (rfm['F'] == 1) & (rfm['R'] >= 4),                        # recent, single purchase
    (rfm['F'] == 1) & (rfm['R'] <= 2) & (rfm['M'] >= 4),      # big spender who went quiet
    (rfm['R'] <= 2),                                          # long inactive
]
segments = ['Loyal (Repeat)', 'New High-Value', 'Recent One-Time',
            'At-Risk High-Value', 'Lost / Inactive']
rfm['segment'] = np.select(conditions, segments, default='Others')

seg = (
    rfm.groupby('segment')
    .agg(customers=('customer_unique_id', 'count'),
         avg_recency=('recency', 'mean'),
         avg_monetary=('monetary', 'mean'),
         total_revenue=('monetary', 'sum'))
    .sort_values('total_revenue', ascending=False)
)
seg['revenue_share_%'] = seg['total_revenue'] / seg['total_revenue'].sum() * 100
display(seg)

fig, ax = plt.subplots(1, 2, figsize=(16, 5))
sns.barplot(x=seg['customers'], y=seg.index, ax=ax[0], orient='h', palette='viridis')
ax[0].set_title('Customers per RFM Segment')
ax[0].set_ylabel('')
sns.barplot(x=seg['revenue_share_%'], y=seg.index, ax=ax[1], orient='h', palette='magma')
ax[1].set_title('Revenue Share % per Segment')
ax[1].set_ylabel('')
save('19_rfm_segments')

# ------------------------------------------------------------
# 7.3 Cohort retention analysis
# Cohort = month of a customer's first purchase
# ------------------------------------------------------------
c = orders_valid[orders_valid['order_status'] == 'delivered'][
    ['customer_unique_id', 'order_purchase_timestamp']].copy()
c['order_month'] = c['order_purchase_timestamp'].dt.to_period('M')
c['cohort'] = c.groupby('customer_unique_id')['order_month'].transform('min')
c['month_index'] = (c['order_month'] - c['cohort']).apply(lambda x: x.n)   # months since first purchase

cohort_counts = (
    c.groupby(['cohort', 'month_index'])['customer_unique_id'].nunique()
    .reset_index()
    .pivot(index='cohort', columns='month_index', values='customer_unique_id')
)
retention = cohort_counts.div(cohort_counts[0], axis=0)       # divide by cohort size (month 0)
retention = retention.loc['2017-01':'2018-03', 0:6]           # complete cohorts, first 6 months

plt.figure(figsize=(11, 7))
sns.heatmap(retention, annot=True, fmt='.1%', cmap='viridis', vmax=0.05,
            cbar_kws={'label': 'Retention'})
plt.title('Monthly Cohort Retention (colour scale capped at 5%)')
plt.xlabel('Months since first purchase')
plt.ylabel('Cohort (first purchase month)')
save('20_cohort_retention')

# ------------------------------------------------------------
# 8. Correlation matrix of numeric columns
# ------------------------------------------------------------
num = delivered[['price', 'freight', 'items_count', 'payment_value', 'installments',
                 'delivery_days', 'estimated_days', 'delay_days', 'review_score']].dropna()

plt.figure(figsize=(9, 7))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
save('21_correlation')

# ------------------------------------------------------------
# 10. Export processed data and download all charts
# ------------------------------------------------------------
orders_df.to_csv('data/processed/orders_master.csv', index=False)
items_df.to_csv('data/processed/items_master.csv', index=False)
rfm.to_csv('data/processed/rfm_segments.csv', index=False)
print('Saved: data/processed/ (orders_master, items_master, rfm_segments)')

!zip -r -q images.zip images/
from google.colab import files
files.download('images.zip')      # use these images in your GitHub README
```
